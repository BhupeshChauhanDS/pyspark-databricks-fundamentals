### 2. withcolumn()  - ADD or change in existing column

df1 ---  changing intger of specific column

df2 ---  changing into column values

df3 --- addinf new column using lit attributes

df4 --- copied a column and creating new one 

```python
from pyspark.sql.functions import col ,lit




data = [
    (1, 'Rahul' , '5000'),
    (2, 'rohan', '4000')
]
schema =['id', 'name', 'salary']

df= spark.createDataFrame(data = data, schema = schema)

df1 = df.withColumn('salary', col('salary').cast('Integer'))             

df2 = df1.withColumn('salary' , col('salary') * 2)

df3 = df2.withColumn('country' , lit('India'))

df4 = df3.withColumn('copiedColumn' , col('salary'))



df3.show()
df3.printSchema()
```

## 4. StructType()  , StructField()  --  inner nested struct schema

```python
from pyspark.sql.types import StructType , StructField , IntegerType , StringType



data = [
    (1, ('Rahul' , 'Kumar') , 5000),
    (2, ('rohan' , 'kumar'), 4000)
]

structName = StructType([\
    StructField('firstname', StringType(), True),\
    StructField('lastname', StringType(), True)])
 


schema =StructType([\
    StructField('id', IntegerType(), True),\
    StructField('name', structName, True),\
    StructField('salary', IntegerType(), True)])




df= spark.createDataFrame(data = data, schema = schema)

df.show()
df.printSchema()
```

## 8. When and otherwise - is like sql if else condition

```python
data  = [(1 , 'Ram', 'M' , '5000') ,  (2 , 'Seeta', 'F' , '20000') , (3 , 'Laksh' ,'' , '10000')]
schema = ['id' , 'name' , 'gender' , 'salary']

df = spark.createDataFrame(data , schema)
df.show()
```

```python
from pyspark.sql.functions import when


df1 = df.select(df.id ,df.name , when(df.gender =='M' , 'Male').when(df.gender =='F' , 'Female').otherwise('Unknown').alias('gender') , df.salary)
df1.show()
```

## 9. Function on columns 
## 
###  alias() -  giving name to any column  \ 
###  sort()\orderBy() - asc() - sorting column \   
###  cast() - changing  datatype\
### like() - same as SQl function to filter data 

```python
data  = [(1 , 'Ram', 'M' , '5000') ,  (2 , 'Seeta', 'F' , '20000') , (3 , 'Laksh' ,'M' , '10000')]
schema = ['id' , 'name' , 'gender' , 'salary']

df = spark.createDataFrame(data , schema)

df1=df.select(df.id.alias('emp_id') ,df.name.alias('emp_name') , df.gender , df.salary)

df2=df1.sort(df1.emp_name.desc())

df3= df2.select(df.emp_id , df.emp_name  , df.gender , df.salary.cast('Int'))

df4=df3.filter(df.emp_name.like('R%'))      # Only those records data where name start with R

df4.show()
df4.printSchema()
```

### filter()  and Where()  -  They both evaluate each row in the DataFrame against your condition, keeping only the rows that return True and discarding the rest.[](url)

```python
data  = [(1 , 'Bhupesh', 'M' , 5000) ,  (2 , 'Shivani', 'F' , 20000) , (3 , 'Rani' ,'F' , 10000)]
schema = ['id' , 'name' , 'gender' , 'salary']

df = spark.createDataFrame(data , schema)

df.show()
```

```python
df1= df.filter(df.gender=='M')
df1.show()
```
