## 7. MapType()  function  -  Bascially its a storing data in dictionalrry type

```python

from pyspark.sql.types import StructType , MapType ,StringType ,StructField

data = [('Bhupesh ' ,  {'hair' : 'Black' , 'eye':'Brown'}) , ('Shivani ' , {'hair' : 'Red' , 'eye':'Black'})]

schema  = StructType([\
    StructField('name' , StringType()) ,\
    StructField('properties' , MapType(StringType() , StringType()))
])


df =spark.createDataFrame(data , schema)

df.show(truncate=False)
display(df)
df.printSchema()
```

### Scenerio 1

```python
from pyspark.sql.functions import col

df1= df.withColumn('hair' , col("properties")['hair'])
df1.show(truncate=False)
```

### Scenerio 2

```python
df2= df1.withColumn('eye' , df.properties.getItem('eye'))
df2.show(truncate=False)
```

### 1. explode()   -- in map function

```python
from pyspark.sql.functions import  explode

df3 =df.select('name' ,'properties' ,explode('properties'))
df3.show()
```

### 2. map_keys -  a column created with onky hold keys

```python
from pyspark.sql.functions import  map_keys

df4 = df.withColumn('keys' , map_keys(df.properties))
df4.show()
```

### 3. map_values-  a column created with onky hold values in map function

```python
from pyspark.sql.functions import  map_values

df5 = df.withColumn('Values' , map_values(df.properties))
df5.show()
```
