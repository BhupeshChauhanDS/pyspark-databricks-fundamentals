## 1. from_json()
## 
### Used to convert json string in to mapType() or StructType()

```python
from pyspark.sql.functions import from_json, col
from pyspark.sql.types import MapType, StringType, IntegerType


# Sample dataset with a JSON string column
data = [
    (1, '{"role": "Data Scientist", "city": "Rewari"}'),
    (2, '{"role": "Data Engineer", "city": "Pune"}')
]

df = spark.createDataFrame(data, ["id", "attributes_json"])

# 1. Define the MapType Schema (KeyType, ValueType)
map_schema = MapType(StringType(), StringType())

# 2. Convert JSON string column to MapType
df_mapped = df.withColumn("attributes_map", from_json(col("attributes_json"), map_schema))

df_mapped.show(truncate=False)
df_mapped.printSchema()
```

```python
from pyspark.sql.functions import from_json, col
from pyspark.sql.types import StructType, StructField, StringType, IntegerType, BooleanType


# Sample data with mixed data types in JSON
data = [
    (1, '{"name": "Bhupesh", "age": 23, "is_active": true}'),
    (2, '{"name": "Shivani", "age": 24, "is_active": false}')
]

df = spark.createDataFrame(data, ["id", "user_json"])

# 1. Define the StructType Schema matching the JSON structure
json_schema = StructType([
    StructField("name", StringType(), True),
    StructField("age", IntegerType(), True),
    StructField("is_active", BooleanType(), True)
])

# 2. Parse JSON string using from_json with StructType
df_parsed = df.withColumn("user_struct", from_json(col("user_json"), json_schema))

df_parsed.show(truncate=False)
df_parsed.printSchema()
```

## 3. json_tuple()
## 

### **function used to extract multiple top-level fields from a JSON string column in a single pass.**

```python
data = [
    (1, '{"name": "Bhupesh", "age": 23, "is_active": true}'),
    (2, '{"name": "Shivani", "age": 24, "is_active": false}')
]

schema = ['id' , 'property']

df = spark.createDataFrame(data, schema)

df.show(truncate=False)
print(df.schema)
```

```python
from pyspark.sql.functions import json_tuple , col

df.select('id' , json_tuple(col('property') , 'name' , 'age' , 'is_active').alias('name' , 'age' , 'is_active') ).show()
```

## 4. get_json_object()
## 


### It extracts a single JSON element or field from a JSON string based on a specified JSONPath expression.


```python
from pyspark.sql.functions import get_json_object, col


# Sample data with nested JSON structure
data = [
    (1, '{"user": {"name": "Bhupesh", "age": 23}, "skills": ["Python", "PySpark"]}'),
    (2, '{"user": {"name": "Shivani", "age": 24}, "skills": ["SQL", "Java"]}')
]

df = spark.createDataFrame(data, ["id", "data_json"])

# Extract single fields using JSONPath
df_extracted = df.select(
    col("id"),
    # Extract nested string field
    get_json_object(col("data_json"), "$.user.name").alias("name"),
    # Extract nested numeric field (returns as String)
    get_json_object(col("data_json"), "$.user.age").cast("integer").alias("age"),
    # Extract specific array element
    get_json_object(col("data_json"), "$.skills[0]").alias("primary_skill")
)

df_extracted.show()
```
